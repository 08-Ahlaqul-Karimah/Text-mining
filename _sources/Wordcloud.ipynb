# SELURUH DATA

```python
# Total Data
datatot = pd.read_csv("https://raw.githubusercontent.com/08-Ahlaqul-Karimah/Pariwisata-New/main/TotalData.csv", delimiter=";", encoding='windows-1252')
datatot.head()
```

```python
datatot['Sentimen'].value_counts()
```

## WORDCLOUD

### Positif


```python
postifdatatot=datatot.loc[datatot["Sentimen"] == "Positif"]
```

```python
# Menggabungkan semua kata dalam satu string dengan spasi sebagai pemisah
positifWords=""
for i in postifdatatot['komentar']:
  positifWords=positifWords +" "+i
# Membuat WordCloud dengan menggunakan string yang telah digabungkan
positifwordCloud = WordCloud(colormap="Blues", width=1600, height=800, background_color="lightcoral", random_state=30, max_font_size=200, min_font_size=20).generate(positifWords)
# Menampilkan WordCloud
plt.imshow(positifwordCloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```

### Negatif

```python
negatifdatatot=datatot.loc[datatot["Sentimen"] == "Negatif"]
```

```python
# Menggabungkan semua kata dalam satu string dengan spasi sebagai pemisah
negatifWords=""
for i in negatifdatatot['komentar']:
  negatifWords=negatifWords +" "+(i)
# Membuat WordCloud dengan menggunakan string yang telah digabungkan
negatifwordCloud = WordCloud(colormap="Reds", width=1600, height=800, background_color="darkgreen", random_state=30, max_font_size=200, min_font_size=20).generate(negatifWords)
# Menampilkan WordCloud
plt.imshow(negatifwordCloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```

### Netral

```python
netraldatatot=datatot.loc[datatot["Sentimen"] == "Netral"]
```

```python
# Menggabungkan semua kata dalam satu string dengan spasi sebagai pemisah
netralWords=""
for i in netraldatatot['komentar']:
  netralWords=netralWords +" "+i
# Membuat WordCloud dengan menggunakan string yang telah digabungkan
netralwordCloud = WordCloud(colormap="rainbow", collocations=True, width=1600, height=800, background_color="white", random_state=30, max_font_size=200, min_font_size=20).generate(netralWords)

# Menampilkan WordCloud
plt.imshow(netralwordCloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
